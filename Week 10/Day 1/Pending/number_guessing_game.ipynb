# Number Guessing Game

Guess the secret number between 1 and 100. You have seven valid attempts. Run the setup cell first, then run the game cell.

```python
import random


def number_guessing_game():
    random_number = random.randint(1, 100)
    max_attempts = 7

    print("I have picked a number between 1 and 100.")

    for attempt in range(1, max_attempts + 1):
        while True:
            try:
                guess = int(input(f"Attempt {attempt}/{max_attempts} - Enter your guess: "))
            except ValueError:
                print("Please enter a whole number.")
                continue

            if not 1 <= guess <= 100:
                print("Your guess must be between 1 and 100.")
                continue
            break

        if guess < random_number:
            print("Too low!")
        elif guess > random_number:
            print("Too high!")
        else:
            print(f"Congratulations! You guessed the number in {attempt} attempt(s).")
            break
    else:
        print(f"Out of attempts! The number was {random_number}.")
```

```python
number_guessing_game()
```
